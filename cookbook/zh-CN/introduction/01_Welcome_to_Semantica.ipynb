Semantica 是一个**语义智能与知识工程框架**。它可以帮助你:

- 从非结构化和半结构化数据构建**知识图谱**
- 在各种异构数据源之上创建统一的**语义层**
- 用结构化知识驱动 **GraphRAG**、AI 智能体和多智能体系统
- 在你的应用中融入**时态感知与质量感知的推理**

### 核心能力

- **通用接入**:文件、网页、订阅源、数据库、代码仓库、数据流
- **丰富的解析**:PDF、Office 文档、HTML、JSON、CSV、图像、代码
- **规范化**:清洗、语言检测、实体规范化、日期/数字标准化
- **语义抽取**:命名实体、关系、事件、语义网络
- **知识图谱构建**:基于实体和关系构建属性图
- **嵌入与向量搜索**:文本与图嵌入、混合检索
- **推理与本体**:基于规则的推理、本体生成与校验
- **可视化与分析**:图可视化与质量指标

## Semantica 适合谁?

- **AI/ML 工程师**:构建需要长期记忆的 GraphRAG 系统、智能体和工具
- **数据工程师**:在大规模异构数据集上编排语义增强管线
- **知识工程师与本体专家**:设计并维护形式化的知识结构
- **研究人员与分析师**:基于文档和数据订阅源创建领域知识图谱
- **产品与平台团队**:将语义智能嵌入到应用和服务中

## 架构概览

Semantica 由三个概念层和多个具体模块组成。

### 层次

- **输入层**
  - 连接文件、网页、API、数据库、邮件、订阅源、代码仓库和数据流
  - 将这些不同的来源规范化为统一的内部表示

- **语义层**
  - 执行解析、清洗、语义抽取、图构建、嵌入和推理
  - 这是**非结构化数据变成结构化知识**的地方

- **输出层**
  - 对外提供知识图谱、嵌入、本体和分析结果
  - 与向量库、图数据库和下游应用集成

## 🧩 Semantica 模块参考

Semantica 在设计上是模块化的。以下是按功能分组的所有可用模块的完整指南。

### 📥 接入与解析
处理原始数据输入与结构的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`ingest`** | **数据接入**<br>连接数据源。 | • 文件、Web、订阅源、数据流接入<br>• 支持数据库、邮件、代码仓库、MCP |
| **`parse`** | **文档解析**<br>将原始内容解析为结构。 | • PDF、HTML、JSON、CSV、Excel<br>• 图像与代码解析 |

### ⚙️ 数据处理
清洗、规范化和切分数据的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`normalize`** | **数据规范化**<br>清洗并标准化文本。 | • 文本清洗与语言检测<br>• 实体、日期、数字规范化 |
| **`split`** | **分块**<br>为 RAG 切分文档。 | • 递归字符切分<br>• 语义与基于 Token 的切分 |

### 🧠 抽取与增强
从原始数据中抽取语义、结构和向量的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`semantic_extract`** | **信息抽取**<br>抽取实体和关系。 | • NER 与关系抽取<br>• 事件与语义网络检测 |
| **`context`** | **智能体记忆**<br>管理 AI 智能体的状态。 | • 长期记忆与历史<br>• 上下文图与 RAG 集成 |

### 🕸️ 知识图谱核心
用于构建、精炼和消解知识图谱的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`kg`** | **图构建**<br>构建并分析图。 | • 图构建与分析<br>• 校验与实体消解 |
| **`conflicts`** | **冲突解决**<br>解决数据矛盾。 | • 来源可靠性评分<br>• 真值发现算法 |
| **`deduplication`** | **实体消解**<br>合并重复实体。 | • 基于相似度的分块(blocking)<br>• 聚类与规范化 |

### 💾 存储与检索
用于持久化和查询数据的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`embeddings`** | **向量嵌入**<br>生成语义向量。 | • 文本与图嵌入<br>• 多提供商支持(OpenAI 等) |
| **`vector_store`** | **向量数据库**<br>存储并搜索向量。 | • 相似度搜索与过滤<br>• 混合搜索(向量 + 关键词) |
| **`graph_store`** | **属性图存储**<br>持久化图数据。 | • Neo4j、FalkorDB 适配器<br>• 支持 Cypher 查询 |
| **`triplet_store`** | **RDF 存储**<br>持久化语义三元组。 | • SPARQL 端点<br>• BlazeGraph、Jena、Virtuoso 适配器 |

### 🔎 推理与分析
用于推导新知识和评估质量的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`reasoning`** | **推理门面**<br>统一的推理接口。 | • Datalog/基于规则的推理<br>• 前向/后向链 |
| **`ontology`** | **本体管理**<br>管理模式与定义。 | • 从数据生成本体<br>• 校验与演化 |
| **`visualization`** | **可视化分析**<br>可视化图与指标。 | • 2D/3D 图可视化<br>• 交互式图表与仪表盘 |
| **`evals`** | **评估**<br>对管线质量做基准测试。 | • RAG 与图质量指标<br>• 与真值(ground truth)对比 |

### 🛠️ 编排与工具
用于管理框架和工作流的模块。

| 模块 | 描述 | 关键能力 |
| :--- | :--- | :--- |
| **`core`** | **框架核心**<br>主入口与配置。 | • 生命周期管理<br>• 插件系统与配置 |
| **`pipeline`** | **工作流编排**<br>管理复杂流程。 | • DAG 执行与重试<br>• 错误处理与可观测性 |
| **`seed`** | **数据种子**<br>初始化知识库。 | • 分类法与本体种子<br>• 参考数据加载 |
| **`export`** | **数据导出**<br>将数据导出为文件。 | • JSON、CSV、RDF、GEXF 导出<br>• 报告生成 |
| **`utils`** | **工具**<br>通用辅助函数。 | • 日志、异步、哈希<br>• 文本处理辅助工具 |

## 核心概念(概览)

- **知识图谱**
  - 节点表示实体,如人物、组织、地点、事件或概念
  - 边表示关系,如 `works_for`、`located_in`、`founded_by`
  - 属性记录特性和元数据,如时间戳、来源和置信度

- **实体与关系**
  - 实体通过 NER 从文本和数据中抽取
  - 关系连接实体,可通过基于模式、基于模型或基于 LLM 的方法抽取

- **嵌入(Embeddings)**
  - 编码文本或图结构语义的数值向量
  - 用于语义搜索、聚类和基于相似度的检索

- **GraphRAG**
  - 将向量搜索与图遍历相结合
  - 同时利用嵌入和图结构,检索丰富的、具备上下文感知的信息

- **本体(Ontology)**
  - 对某一领域中类、关系和约束的形式化模型
  - 用于标准化语义、支持推理并整合异构数据

- **质量与治理**
  - 质量指标(完整性、一致性、准确性、覆盖度)
  - 在知识图谱层面进行冲突检测与解决

````python
import os
from pathlib import Path

os.environ["SEMANTICA_API_KEY"] = "your_openai_key"
os.environ["SEMANTICA_EMBEDDING_PROVIDER"] = "openai"
os.environ["SEMANTICA_MODEL_NAME"] = "gpt-4"

config_text = """api_keys:
  openai: your_key_here
  anthropic: your_key_here
embedding:
  provider: openai
  model: text-embedding-3-large
  dimensions: 3072
knowledge_graph:
  backend: networkx
  temporal: true
"""
Path("config.yaml").write_text(config_text, encoding="utf-8")
Path("config.yaml").read_text(encoding="utf-8")
````

## 最小化端到端管线

下一个示例展示如何依次显式地使用多个模块,这与前面讨论的架构一致:

1. 接入一个文档目录
2. 将它们解析为结构化文档
3. 规范化文本
4. 抽取实体与关系
5. 构建并分析知识图谱
6. 创建嵌入并存入向量库
7. 运行一次混合语义搜索查询

````python
from semantica.ingest import FileIngestor
from semantica.parse import DocumentParser
from semantica.normalize import TextNormalizer
from semantica.semantic_extract import NERExtractor, RelationExtractor
from semantica.kg import GraphBuilder, GraphAnalyzer
from semantica.embeddings import EmbeddingGenerator
from semantica.vector_store import VectorStore, HybridSearch

ingestor = FileIngestor()
documents = ingestor.ingest(str(docs_dir))

parser = DocumentParser()
parsed_docs = parser.parse(documents)

normalizer = TextNormalizer()
normalized_docs = normalizer.normalize(parsed_docs)

ner = NERExtractor()
entities = ner.extract(normalized_docs)
rel_extractor = RelationExtractor()
relationships = rel_extractor.extract(normalized_docs, entities)

builder = GraphBuilder()
kg = builder.build(entities, relationships)
analyzer = GraphAnalyzer()
metrics = analyzer.analyze(kg)

emb_generator = EmbeddingGenerator()
embeddings = emb_generator.generate_embeddings(documents, data_type="text")

vec_store = VectorStore()
vec_store.store(embeddings, documents, metadata={})
hybrid = HybridSearch(vec_store)
search_results = hybrid.search("Apple founders", top_k=3)
len(search_results)
````

## 本体生成

你还可以根据抽取出的实体和关系自动生成本体(对你所在领域的形式化模型)。

````python
from semantica.ontology import OntologyGenerator

generator = OntologyGenerator(base_uri="https://example.org/ontology/")

# Generate ontology from the extracted data
ontology = generator.generate_ontology({
    "entities": entities,
    "relationships": relationships
})

# View inferred classes
[cls["name"] for cls in ontology.get("classes", [])[:5]]
````

## 进阶:数据切分与分块

对于 RAG 应用,把文档切分成更小的分块至关重要。Semantica 为此提供了 `split` 模块。

````python
from semantica.split import TextSplitter

splitter = TextSplitter(chunk_size=100, chunk_overlap=20)
chunks = splitter.split_documents(documents)

print(f"Original documents: {len(documents)}")
print(f"Generated chunks: {len(chunks)}")
````

## 进阶:推理与推断

`reasoning` 模块允许你使用逻辑规则从已有知识中推导出新事实。

````python
from semantica.reasoning import Reasoner

# Simple rule: If X founded Y, then X works_for Y
rule = """
IF (?x founded ?y) THEN (?x works_for ?y)
"""

reasoner = Reasoner()
reasoner.add_rule(rule)
inferred_facts = reasoner.infer_facts(kg)

print(f"Inferred {len(inferred_facts)} new facts")
````
